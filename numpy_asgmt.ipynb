```python
d = np.array([100,23,45,67,89,98,87,65,43,32,56,78])

mask = (d > 80)

print(mask)
print(d[mask])
print(d[~mask])
```

```python
d = np.array([100,23,45,67,89,98,87,65,43,32,56,78])

a = (d > 70) & (d < 90)

print(d[a])
```

```python
print(np.where(d > 70))
```

```python
cond = [
    ((d >= 40) & (d <= 75)),
    ((d > 75) & (d <= 100))
]

choice = ["Small", "Medium"]

print(np.select(cond, choice, default="Large"))
```

```python
m = (d % 2 == 0) | (d % 5 == 0) | (d % 7 == 0)

data = d[~m]

print(data)
```
